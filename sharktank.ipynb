```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

shark = pd.read_csv("dataset_shark.csv")
shark.head()
```

```python
print("Rows:", shark.shape[0])
print("Columns:", shark.shape[1])

shark.info()
```

```python
shark.isnull().sum()
```

```python
shark.describe()
```

```python
shark_clean = shark.copy()

shark_clean.columns = (
    shark_clean.columns
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
)

shark_clean["equity"] = (
    shark_clean["equity"]
    .astype(str)
    .str.replace("%", "", regex=False)
    .str.strip()
    .replace("nan", np.nan)
)

shark_clean["equity"] = pd.to_numeric(
    shark_clean["equity"], errors="coerce"
)

# Convert investor columns to 1/0
investors = ["anupam", "ashneer", "namita", "aman",
             "peyush", "vineeta", "ghazal"]

for investor in investors:
    shark_clean[investor] = (
        shark_clean[investor]
        .astype(str)
        .str.strip()
        .str.upper()
        .map({"Y": 1, "N": 0})
    )

# Remove duplicate rows
shark_clean = shark_clean.drop_duplicates()

print("Rows after cleaning:", shark_clean.shape[0])
print("Columns:", shark_clean.shape[1])

shark_clean.head()
```

```python
for investor in investors:
    shark_clean[investor] = (
        shark_clean[investor]
        .astype(str)
        .str.strip()
        .str.upper()
        .map({"Y": 1, "N": 0})
    )

# Remove duplicate rows
shark_clean = shark_clean.drop_duplicates()

print("Rows after cleaning:", shark_clean.shape[0])
print("Columns:", shark_clean.shape[1])

shark_clean.head()
```

```python
print(shark_clean.columns.tolist())
```

```python
investment_col = "investment_amount_in_lakhs_inr_"

shark_clean["deal_status"] = np.where(
    shark_clean[investment_col] > 0,
    "Deal",
    "No Deal"
)

print(shark_clean["deal_status"].value_counts())
```

```python
print("Total startups:", len(shark_clean))

print("Total investment:",
      shark_clean[investment_col].sum(),
      "Lakhs INR")

print("Total investment:",
      shark_clean[investment_col].sum() / 100,
      "Crores INR")

print("\nAverage investment:",
      shark_clean[investment_col].mean(),
      "Lakhs INR")
```

```python
print("\nInvestor participation:")

for investor in investors:
    print(
        investor.title(),
        ":",
        shark_clean[investor].sum(),
        "deals"
    )
```

```python
investor_totals = (
    shark_clean[investors]
    .sum()
    .sort_values(ascending=False)
)

print(investor_totals)
```

```python
investor_investment = {}

for investor in investors:
    investor_investment[investor.title()] = shark_clean.loc[
        shark_clean[investor] == 1,
        investment_col
    ].sum()

investor_investment = pd.Series(investor_investment).sort_values(
    ascending=False
)

print(investor_investment)
```

```python
top_startups = shark_clean[
    ["brand", "idea", investment_col, "equity"]
].sort_values(
    investment_col,
    ascending=False
).head(10)

top_startups
```

```python
correlation = shark_clean[
    [investment_col, "equity"]
].corr()

print(correlation)
```

```python
season_analysis = shark_clean.groupby("season").agg(
    startups=("brand", "count"),
    total_investment=(investment_col, "sum"),
    average_investment=(investment_col, "mean")
).reset_index()

season_analysis
```

```python
output_file = "cleaned_shark_tank_data.xlsx"

shark_clean.to_excel(output_file, index=False)

print("Excel file created:", output_file)
```

```python
output_file = "cleaned_shark_tank_data.xlsx"

shark_clean.to_excel(output_file, index=False)

print("Excel file created:", output_file)
```

```python
with pd.ExcelWriter("shark_tank_analysis.xlsx") as writer:

    shark_clean.to_excel(
        writer,
        sheet_name="Cleaned Data",
        index=False
    )

    investor_totals.to_frame("Deal Count").to_excel(
        writer,
        sheet_name="Investor Deals"
    )

    investor_investment.to_frame("Investment Lakhs").to_excel(
        writer,
        sheet_name="Investor Investment"
    )

    top_startups.to_excel(
        writer,
        sheet_name="Top Startups",
        index=False
    )

    season_analysis.to_excel(
        writer,
        sheet_name="Season Analysis",
        index=False
    )

print("Analysis Excel file created successfully!")
```
